# 3-Handling Data

## 3-1 Handl Type of Columns

```python
df.dtypes
```

```python
for col in df.columns:
    print(df[col].value_counts())
```

```python
df['Price']=df['Price'].str.replace(r'[^0-9.]', '', regex=True)
df['Price']=df['Price'].astype('float64')
df['Price']
```

```python
df['Quantity']=df['Quantity'].str.replace(r'[^0-9.]', '', regex=True)
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce').astype('Int64')  # 'Int64' handles NaN
df['Quantity']
```

## 3-2 Handl Data

```python
df['Order_Date']=pd.to_datetime(df['Order_Date'],format='mixed')
df['Order_Date']
```

## 3-3 Handl Missing Values 

```python
df.isnull().sum()
```

```python
df.fillna({'Quantity':int(df['Quantity'].mean())},inplace=True)
df.dropna(subset='Product',axis=0,inplace=True)
df
```

## 3-4 Duplicates 

```python
print(df[df.duplicated()])
```

```python
df.drop_duplicates(inplace = True)
```

```python
df.drop(index=5,axis=1,inplace=True)
```

```python
df.reset_index(drop=True,inplace=True)
```
