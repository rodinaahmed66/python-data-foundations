```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
sns.set(style='whitegrid')
warnings.filterwarnings('ignore')
```

```python
np.random.seed(42)
data = {
'Name':['Mahmoud','Reem','Ahmed','Hany','Mona','Hadeer','Ramy','Sally','David','Mark','Karen','Walid','Mina','Gad','Ayman']
,'Age':np.random.choice([14,15,16],15)
,'ID':np.linspace(0,14,15)
,'Bouns':np.random.randint(5, 10, 15)
,'Mark':np.random.uniform(10, 20, 15).astype(int)
,'Class':[np.random.choice(['A','B','c']) for j in range(0,15)]
}
df=pd.DataFrame(data)
df
```

# 1. Line Plot

```python
plt.figure(figsize=(8,5))
sns.lineplot(x='ID',y='Bouns',data=df,marker='o', linestyle=':',color='orange')
plt.xlabel('ID')
plt.ylabel('Bouns')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.lineplot(x='Name',y='Bouns',data=df,marker='X',markersize=15, linestyle='--',linewidth=4)
plt.xticks(df['Name'], rotation=90)
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.lineplot(x='Name',y='Bouns',data=df,marker='D',markersize=15, linestyle='-',linewidth=4,color='red')
plt.xticks(df['Name'], rotation=90)
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.lineplot(x='Name',y='Bouns',data=df,marker='X',markersize=15, linestyle='--',linewidth=4,label='Bouns')
sns.lineplot(x='Name',y='Age',data=df,marker='X',markersize=15, linestyle='--',linewidth=4,label='Age')
plt.xticks(df['Name'], rotation=90)
plt.legend()
plt.show()
```

# 10. Heatmap

```python
# Create a correlation matrix
corr_mat=df[['Age','Mark','Bouns']].corr()
corr_mat
```

```python
plt.figure(figsize=(8, 5))
sns.heatmap(corr_mat, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Heatmap ')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.heatmap(corr_mat, annot=True, cmap='viridis', vmin=-1, vmax=1)
plt.title('Heatmap ')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.heatmap(corr_mat, annot=True, cmap='magma', vmin=-1, vmax=1,linecolor='black')
plt.title('Heatmap ')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.heatmap(corr_mat, annot=True, cmap='BuGn', vmin=-1, vmax=1)
plt.title('Heatmap ')
plt.show()
```
