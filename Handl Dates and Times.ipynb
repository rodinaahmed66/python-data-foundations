# TimeStep

```python
#  A timestamp is the number of seconds (or milliseconds) elapsed since January 1, 1970, 00:00:00 UTC .
# Example: 1514665153.0 represents Sat Dec 30 2017 15:09:03 UTC. 
# Convert to datetime and print 
ts = 1514665153.0
print(datetime.fromtimestamp(ts)) 
```

# TimeZone

```python
# Eastern Time (UTC-5)
ET = timezone(timedelta(hours=-5))  
dt4 = datetime(2023, 10, 1, 12, 0, tzinfo=ET)
print(dt4)
```

```python
#Convert between time zones using astimezone()
# India Time (UTC+5:30)
IST = timezone(timedelta(hours=5, minutes=30))  
# Converts ET to IST
print(dt4.astimezone(IST))  
```

```python
eastern = tz.gettz('America/New_York')
dt5 = datetime(2023, 3, 12, 2, 30, tzinfo=eastern)  # DST transition
print(dt5)
```
