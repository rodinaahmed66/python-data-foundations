```python
from urllib.request import urlopen,urlretrieve, Request 
from bs4 import BeautifulSoup
import pandas as pd
import requests 
import os
import json
import warnings
warnings.filterwarnings('ignore')
```

```python
# Function to fetch and parse the webpage
def fetch_parse(url):
    
    response=requests.get(url)
    #shows the HTTP status code (200 = success)
    if response.status_code == 200:
        print("Request accepted! Status code:", response.status_code)
        soup=BeautifulSoup(response.content,'html.parser')
        return soup
    else:
        print(f"Request failed with status code: {response.status}")
        return None
```

```python
def Extracted_Data_Text(soup):
    Extracted_Data=''
    for p in soup.find_all('p'):
          Extracted_Data+=p.get_text()
    return Extracted_Data
```

```python
#function to save data as txt file
def save_data_txt(data,dir,file_name):
    os.makedirs(dir,exist_ok=True)
    file_name = os.path.join(dir, file_name)
    with open(file_name, 'w', encoding='utf-8') as file:
        file.write(data)
    print("Data saved to", file_name)
```

```python
# function to print data
def print_data(data):
    print('Extracted Data')
    print(data)
```

```python

# Authentication parameters
token_url = "https://icdaccessmanagement.who.int/connect/token"
ClientId = '819f5036-1850-46eb-92c9-88e091221ca9_e5e115ef-dc00-4e99-b6cc-a91a1dd14156'
ClientSecret = 'A4brv7rJ5hPGkfjNzgHqVAlwzupwdYVZoEFjFuqg6hI='
scope = 'icdapi_access'
grant_type = 'client_credentials'

# Search parameters
search_url = "https://id.who.int/icd/entity/search"
search_query = {"q": "diabetes"}

# Get access token
payload = {
    'client_id': ClientId,
    'client_secret': ClientSecret,
    'scope': scope,
    'grant_type': grant_type
}
token_response = requests.post(token_url, data=payload)
access_token = token_response.json().get("access_token")

# Prepare headers
headers = {
    'Authorization': f'Bearer {access_token}',
    'Accept': 'application/json',
    'Accept-Language': 'en',
    'API-Version': 'v2'
}

# Perform search
search_response = requests.get(search_url, headers=headers, params=search_query)
data = search_response.json()

# Save to JSON file
with open('icd_search_results.json', 'w', encoding='utf-8') as f:
    json.dump(data, f, indent=4, ensure_ascii=False)

# Function to clean HTML tags
def clean_html(raw_html):
    return BeautifulSoup(raw_html, "html.parser").get_text()

# Function to display formatted results
def print_nice_output(data):
    if data.get("error"):
        print("Error:", data.get("errorMessage"))
        return

    print("\nSearch Results for 'Diabetes':")
    print(f"Unique Search ID: {data.get('uniqueSearchId')}")
    print("\nResults:")
    
    for entity in data.get("destinationEntities", []):
        print("\n----------------------------")
        print(f"ID: {entity.get('id')}")
        print(f"Title: {clean_html(entity.get('title', 'N/A'))}")
        print(f"Chapter: {entity.get('chapter', 'N/A')}")
        print(f"ICD Code: {entity.get('theCode', 'N/A')}")
        print(f"Score: {entity.get('score', 'N/A')}")
        print(f"Is Leaf Node: {'Yes' if entity.get('isLeaf') else 'No'}")
        print(f"Title is Search Result: {'Yes' if entity.get('titleIsASearchResult') else 'No'}")
        print(f"Title is Top Score: {'Yes' if entity.get('titleIsTopScore') else 'No'}")

# Display results
print_nice_output(data)
```
