```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
# Numerical vs Numerical Data
# Scatter Plot
# Line Plot
# Histogram

#Numerical vs Categorical Data
# Box Plot
# Scatter
# Bar Plot

#Categorical vs Categorical Data
```

```python
np.random.seed(42)
data = {
'Name':['Mahmoud','Reem','Ahmed','Hany','Mona','Hadeer','Ramy','Sally','David','Mark','Karen','Walid','Mina','Gad','Ayman']
,'Age':np.random.choice([14,15,16],15)
,'ID':np.linspace(0,14,15)
,'Bouns':np.random.randint(5, 10, 15)
,'Mark':np.random.uniform(10, 20, 15).astype(int)
,'Class':[np.random.choice(['A','B','c']) for j in range(0,15)]
}
df=pd.DataFrame(data)
df
```

# 1. Line Plot

```python
plt.figure(figsize=(5,5))
plt.plot(df['ID'],df['Mark'])
plt.xlabel('ID Of Students')
plt.ylabel('Marks')
plt.title('ID VS Marks')
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.plot(df['ID'],df['Mark'],marker='o',color='orange')
plt.fill_between(df['ID'], df['Mark'], color='orange', alpha=0.3)
plt.xlabel('ID Of Students')
plt.ylabel('Marks')
plt.title('ID VS Marks')
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.plot(df['ID'],df['Mark'],marker='^',color='red',markersize=10, linewidth=3)
plt.xlabel('ID Of Students')
plt.ylabel('Marks')
plt.title('ID VS Marks')
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.plot(df['ID'],df['Mark'],marker='^',linestyle='-.',color='g',markersize=10, linewidth=3)
plt.xlabel('ID Of Students')
plt.ylabel('Marks')
plt.title('ID VS Marks')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(df['ID'],df['Mark'],color='y',markersize=10, linewidth=3,label='Mark')
plt.xlabel('ID Of Students')
plt.ylabel('Value')
plt.plot(df['ID'],df['Bouns'],color='c',markersize=10, linewidth=3,label='Bouns')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.plot(df['ID'],df['Mark'],color='g',marker='x',linestyle='-',markersize=10, linewidth=3,alpha=0.6,label='Mark')
plt.xlabel('ID Of Each Student')
plt.ylabel('Marks')
plt.twinx()
plt.plot(df['ID'],df['Bouns'],color='m',marker='*',linestyle=':',markersize=10, linewidth=3,alpha=0.6,label='Bouns')
plt.ylabel('Bouns')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(df['ID'],df['Mark'],color='red',markersize=10, linewidth=3,label='Mark')
plt.tick_params('y', colors='red')
plt.xlabel('ID Of Each Student')
plt.ylabel('Marks')
plt.twinx()
plt.plot(df['ID'],df['Bouns'],color='blue',markersize=10, linewidth=3,label='Bouns')
plt.tick_params('y', colors='blue')
plt.ylabel('Bouns')
plt.legend()
plt.show()
```

```python
# plt.annotate(
#     text,          Text to display
#     xy,            (x,y) coordinate to annotate
#     xytext,        (x,y) coordinate for text placement
#     arrowprops,    Arrow styling (optional)
#     ha             Horizontal/vertical alignment
#     **kwargs       Text styling (fontsize, color, etc.)
# )

plt.figure(figsize=(8, 5))
plt.plot(df['ID'], df['Mark'], marker='o', linestyle='-', color='green', 
         markersize=8, linewidth=2, alpha=0.8)
for i, txt in enumerate(df['Mark']):
    plt.annotate(txt, (df['ID'][i], df['Mark'][i]), 
                 textcoords="offset points", xytext=(0,20), ha='center')
plt.xlabel('ID')
plt.ylabel('Mark')
plt.title('Line Plot with Annotations')
plt.grid(True)
plt.show()
```

# 6. Box Plot

```python
plt.figure(figsize=(10, 6))
data=df.select_dtypes('number').drop(columns='ID')
plt.boxplot(data,labels=data.columns,patch_artist=True,boxprops=dict(facecolor='lightgreen', color='black'),
            whiskerprops=dict(color='blue',linewidth=1.5), 
            medianprops=dict(color='red',linewidth=1.5),
            capprops=dict(color='blue',linewidth=1.5),
            flierprops=dict(marker='o', color='orange', markersize=5))
plt.title('Boxplot of Age, Bouns, and Mark')
plt.ylabel('Values')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
grouped = df.groupby(['Class'])['Mark'].apply(list).reset_index()
grouped
```

```python
plt.figure(figsize=(10, 6))
plt.boxplot(list(grouped['Mark']),labels=grouped['Class'],patch_artist=True,boxprops=dict(facecolor='lightpink', color='black'),
            whiskerprops=dict(color='black'),
            medianprops=dict(color='black'),
            flierprops=dict(markerfacecolor='red', marker='o'))
plt.title('Boxplot of Marks in Each Class')
plt.ylabel('Marks')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
grouped = df.groupby(['Age'])['Mark'].apply(list).reset_index()
grouped
```

```python
plt.figure(figsize=(10, 6))
plt.boxplot(list(grouped['Mark']),labels=grouped['Age'],patch_artist=True,boxprops=dict(facecolor='purple', color='black'),
            whiskerprops=dict(color='orange', linewidth=1.5),
            capprops=dict(color='blue', linewidth=2),
            medianprops=dict(color='black', linewidth=1.5),
            flierprops=dict(marker='o', color='red', markersize=5))
plt.title('Boxplot of Marks in Each Class')
plt.ylabel('Marks')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.boxplot(list(grouped['Mark']),labels=grouped['Age'],notch=True, patch_artist=True,boxprops=dict(facecolor='lightblue', color='black'),
            medianprops=dict(color='black'),
            whiskerprops=dict(color='black'))
plt.title('Boxplot of Marks in Each Class')
plt.ylabel('Marks')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

# 7. Pie Chart

```python
plt.figure(figsize=(8, 5))
plt.pie(df['Age'].unique(),labels=df['Age'].unique(),autopct="%1.1f%%", startangle=90,colors=sns.diverging_palette(250, 0))
plt.title('Pie Chart of Ages')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.pie(df['Age'].unique(),labels=df['Age'].unique(),autopct="%1.1f%%", startangle=90,explode=[0.1, 0, 0.05])
plt.title('Pie Chart of Ages with explode')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.pie(df['Age'].unique(),labels=df['Age'].unique(), startangle=90,colors=sns.color_palette('coolwarm'),wedgeprops=dict(width=0.5))
plt.title('Pie Chart of Ages')
plt.show()
```
