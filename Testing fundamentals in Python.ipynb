# 1.DocString Testing

```python
import doctest
def add(n, m):
    '''
    >>> add(2, 3)
    5
    >>> add(-1, 1)
    5
    >>> add(0, 0)
    0
    '''
    return n + m
doctest.testmod()
```

```python
def number_of_even(n):
    '''
    >>> number_of_even(10)
    6
    >>> number_of_even(5)
    3
    >>> number_of_even(0)
    1
    '''
    counter=0
    for i in range(n):
        if(i%2==0):
            counter+=1
    return counter

if __name__ == "__main__":
    doctest.testmod()
```

```python
import unittest

def prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
           
class TestMathOperations(unittest.TestCase):

    def test_add(self):
        self.assertFalse(prime(9)) 
        self.assertTrue(prime(997))
        self.assertFalse(prime(0))
        self.assertFalse(prime(1)) 

suite = unittest.TestLoader().loadTestsFromTestCase(TestMathOperations)
unittest.TextTestRunner().run(suite)
```

```python
import unittest

def Palindrome(s):
    for i in range(0,len(s),1):
        if(s[i]!=s[len(s)-i-1]):    
                return 'Not Palindrome'
    return 'Palindrome'

class TestMathOperations(unittest.TestCase):
    
    def setUp(self):
                self.test_cases = [
            ('anna', 'Palindrome'),
            ('reem', 'Not Palindrome'),
            ('abcba', 'Palindrome'),
            ('mn', 'Not Palindrome'),
            ('123', 'Not Palindrome'),
            ('1221', 'Palindrome')
        ]

    def test_palindrome(self):
        for i,j in self.test_cases:
            self.assertEqual(Palindrome(i),j)
    
    def tearDown(self):
        self.test_cases.clear()
        

if __name__ == '__main__':
    unittest.main(argv=['first-arg-is-ignored'], exit=False)
```

```python
import pytest

def count(s):
    c = {}
    for i in s:
        c[i] = c.get(i, 0) + 1
    return c

@pytest.mark.parametrize("input_string, expected_output", [
    ("anna", {'a': 2, 'n': 2}),
    ("", {}),
    ("abc", {'a': 1, 'b': 1, 'c': 1}),
    ("aaa", {'a': 3}),
    ("Aa", {'A': 1, 'a': 1}),
    ("123321", {'1': 2, '2': 2, '3': 2}),
])

def test_count(input_string, expected_output):
    assert count(input_string) == expected_output

#run ---> in terminal -->PS C:\Users\ASUS> cd R:/
#----->PS R:\> pytest filename.py 
```

```python
import pytest
def decorator(func):
    def wrapper(arg):
      result=func(arg)/len(arg)
      print(f"Average of elements is {result}")
      return result
    return wrapper

@decorator
def summ(arg):
    return sum(arg)

def test_positive(capsys):
    result = summ([1, 2, 3])
    captured = capsys.readouterr()
    assert result == 6  
    assert captured.out.strip() == "Average of elements is 2.0"  

#to skip this test
@pytest.mark.skip
def test_positive(capsys):
    result = summ([-1, -2, -3])
    captured = capsys.readouterr()
    assert result == -6  
    assert captured.out.strip() == "Average of elements is -2.0"  # Verify the printed average

#run ---> in terminal -->PS C:\Users\ASUS> cd R:/
#----->PS R:\> pytest filename.py 
```
