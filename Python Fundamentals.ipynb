```python
def identify_type(value):
    match value:
        case int():
            return "It's an integer."
        case str():
            return "It's a string."
        case float():
            return "It's a float."
        case _:
            return "It's an unknown type."

print(identify_type(100))
print(identify_type("hello"))
print(identify_type(6.5))
```

```python
#generator with function
def count_up_to(n):
      count = 1
      while count <= n:
          yield count
          count += 1

counter = count_up_to(3)
print(next(counter)) 
print(next(counter)) 
print(next(counter)) 
```

## 5.6 Decorators

```python
def my_decorator(func):
    def wrapper():
        print("🚀 قبل تنفيذ الدالة")
        func()  # استدعاء الدالة الأصلية
        print("✅ بعد تنفيذ الدالة")
    return wrapper

@my_decorator
def say_hello():
    print("👋 Hello, World!")

say_hello()
```

```python
#decorators
def my_decorator(func):
    def wrapper(a,b):
        print("decorator is start")
        result = func(2*a,2*b) #call original function
        print("decorator is end")
        return result
    return wrapper

@my_decorator
def multiply(a,b):
    return a*b

multiply(3,4)
```

```python
#two decorators

#make decorator 1 to sum a and b
def decorator1(func):
    def wrapper(a,b): 
       print('hello from decorator 1')
       func(a,b)
       result=a+b
       print('sum of two numbers is ',result)
       print('by from decorator 1')
    return wrapper


#make decorator 2 to sub a and b
def decorator2(func):
    def wrapper(a,b): 
       func(a,b)
       print('hello from decorator 2')
       print('there is two numbers',func(a,b))
       result=a-b
       print('subs of two numbers is ',result)
       print('by from decorator 2')
    return wrapper

@decorator1
@decorator2  
def number(a,b):
    return a,b

number(2,7)
```

## 6.2. Handling Multiple Exceptions

```python
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError as z:
        print(f"Error: {z}")
    else:
        print(f"Result: {result}")
    finally:
        print("Execution completed.")

# Test cases
divide(10, 2)
divide(10, 0)
```

## 8.3 Inheritance

```python
# Superclass
class Vehicle:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def info(self):
        return f"{self.year} {self.make} {self.model}"

# Subclass
class Car(Vehicle):
    def __init__(self, make, model, year, doors):
        Vehicle.__init__(self,make, model, year)  
        self.doors = doors

    def information(self):
        return f"{Vehicle.info(self)} with {self.doors} doors"


my_car = Car("Toyota", "Camry", 2020, 4)
print(my_car.info())  
print(my_car.information())
```

## 8.4 Polymorphism

## 8.6 Descriptors

```python
class BankAccount:
  def __init__(self, balance):
    self.balance = balance

  @property
  def balance(self):
    return f"${round(self._balance, 2)}"

  @balance.setter
  def balance(self, new_balance):
    if new_balance > 0:
      self._balance = new_balance

  @balance.deleter
  def balance(self):
    print("Deleting the 'balance' attribute")
    del self._balance

a1=BankAccount(5531429)
print(a1.balance)
a1.balance = 150
print(a1.balance)
del a1.balance
```

## 8.7 Attribute Access

```python
class Dog:
    def __init__(self, name):
        self.name = name
    
    def __getattr__(self, attr):
        if attr == "bark":
            return "Woof!"
        raise AttributeError(f"'Dog' object has no attribute '{attr}'")

buddy = Dog("Buddy")
print(buddy.bark)  # Output: "Woof!"
#print(buddy.age)   # Raises AttributeError
```

```python
class Dog:
    def __setattr__(self, attr, value):
        if attr == "age" and value < 0:
            raise ValueError("Age cannot be negative")
        super().__setattr__(attr, value)

buddy = Dog()
buddy.age = 5     # Works
#buddy.age = -1    # Raises ValueError
```

```python
class DynamicAttributes:
    def __init__(self):
        self._data = {}

    def __getattr__(self, name):
        if name in self._data:
            return self._data[name]
        raise AttributeError(f"'{name}' not found")

    def __setattr__(self, name, value):
        if name == "_data":
            super().__setattr__(name, value)
        else:
            self._data[name] = value

# Usage
obj = DynamicAttributes()
obj.x = 10         
print(obj.x)  
print(obj._data)      
#print(obj.y)  ---> attribytte error      
```

## 9.1 Text Files

```python
# "r"	Read (default) – file must exist
# "w"	Write – creates a new file or overwrites
# "a"	Append – adds content at the end
# "r+"	Read & write (does not overwrite)
# "w+"	Read & write (overwrites)
# "a+"	Read & append
# "x"	Create a new file (raises error if exists)
path='example.txt'
with open(path, "r") as file:
    content = file.read()
    print(content)

with open(path, "w") as file:
    file.write('hi from code')

with open(path, "r") as file:
    print(file.read())

with open(path, "a+") as file:
    file.write('\nhello')
    print(file.readlines())  
```
